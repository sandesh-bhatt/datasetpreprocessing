```python
import numpy as np

url = "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz"
!wget -q "$url" -O mnist.npz

with np.load("mnist.npz") as data:
    X_train = data["x_train"]
    y_train = data["y_train"]

print(X_train.min(), X_train.max())
# Expected: 0 255
```

```python
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter

img = Image.fromarray(X_train[0])

radii = [0, 1, 2, 3, 5, 7, 10]

fig, axes = plt.subplots(2, 4, figsize=(12, 6))

for ax, radius in zip(axes.flat, radii):
    blurred = (
        img if radius == 0
        else img.filter(ImageFilter.GaussianBlur(radius))
    )

    ax.imshow(blurred, cmap="gray", vmin=0, vmax=255)
    ax.set_title("Original" if radius == 0 else f"Radius = {radius}")
    ax.axis("off")

axes.flat[-1].axis("off")
plt.tight_layout()
plt.show()
```
