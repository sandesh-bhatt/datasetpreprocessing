```python
import numpy as np
import cv2
import matplotlib.pyplot as plt

data = np.load("mnist.npz")

x_train = data['x_train']
y_train = data['y_train']

x_test = data['x_test']
y_test = data['y_test']

print("Original shape:", x_train.shape)
print("Original data type:", x_train.dtype)
```

```python
def preprocess_image(img):
    # 1. Resize image
    img = cv2.resize(img, (20, 20), interpolation=cv2.INTER_AREA)

    # 2. Gaussian blur to reduce noise
    img = cv2.GaussianBlur(img, (3, 3), 0)

    # 3. Resize back to 28x28
    img = cv2.resize(img, (28, 28), interpolation=cv2.INTER_LINEAR)

    # 4. Normalize pixel values to [0, 1]
    img = img.astype(np.float32) / 255.0

    return img
```

```python
x_train_processed = np.array([
    preprocess_image(img) for img in x_train
])

x_test_processed = np.array([
    preprocess_image(img) for img in x_test
])

print("Processed training shape:", x_train_processed.shape)
print("Processed testing shape:", x_test_processed.shape)
```

```python
plt.figure(figsize=(10, 4))

for i in range(10):
    plt.subplot(2, 10, i + 1)
    plt.imshow(x_train[i], cmap='gray')
    plt.title("Original")
    plt.axis('off')

    plt.subplot(2, 10, i + 11)
    plt.imshow(x_train_processed[i], cmap='gray')
    plt.title("Processed")
    plt.axis('off')

plt.tight_layout()
plt.show()
```
