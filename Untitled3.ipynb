```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
import os
```

```python
url = "https://storage.googleapis.com/tensorflow/tf-keras-datasets/mnist.npz"

!wget -q {url} -O mnist.npz
```

```python
data = np.load("mnist.npz")

x_train = data["x_train"]
y_train = data["y_train"]

x_test = data["x_test"]
y_test = data["y_test"]

print("Dataset loaded successfully")
```

```python
print("Null values in x_train:", pd.isnull(x_train).sum())
print("Null values in y_train:", pd.isnull(y_train).sum())
```

```python
print("Training images:", x_train.shape)
print("Training labels:", y_train.shape)
print("Testing images:", x_test.shape)
print("Testing labels:", y_test.shape)

print("Data type:", x_train.dtype)
print("Minimum:", x_train.min())
print("Maximum:", x_train.max())
```

```python
plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")


plt.show()
```

```python
x_train_norm = x_train.astype(np.float32) / 255.0
x_test_norm = x_test.astype(np.float32) / 255.0

print(x_train_norm.min(), x_train_norm.max())
```

resizing open cv


```python
resized = cv2.resize(x_train[0], (64, 64))

plt.imshow(resized, cmap="gray")
plt.title("Resized Image: 64 × 64")
plt.axis("off")
plt.show()
```

```python
img = x_train[0]

gaussian = cv2.GaussianBlur(img, (3, 3), 0)

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian Blur")

plt.show()
```

```python
canny = cv2.Canny(img, 50, 150)

sobel_x = cv2.Sobel(
    img, cv2.CV_64F, 1, 0, ksize=3
)
sobel_y = cv2.Sobel(
    img, cv2.CV_64F, 0, 1, ksize=3
)

plt.figure(figsize=(10, 3))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")

plt.subplot(1, 3, 2)
plt.imshow(canny, cmap="gray")
plt.title("Canny")

plt.subplot(1, 3, 3)
plt.imshow(np.hypot(sobel_x, sobel_y), cmap="gray")
plt.title("Sobel")

plt.show()
```

```python
matrix = cv2.getRotationMatrix2D((14, 14), 15, 1)
rotated = cv2.warpAffine(img, matrix, (28, 28))

matrix = np.float32([[1, 0, 3], [0, 1, 3]])
shifted = cv2.warpAffine(img, matrix, (28, 28))

plt.figure(figsize=(8, 3))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")

plt.subplot(1, 3, 2)
plt.imshow(rotated, cmap="gray")
plt.title("Rotated")

plt.subplot(1, 3, 3)
plt.imshow(shifted, cmap="gray")
plt.title("Shifted")

plt.show()
```

```python
inverted = 255 - img

plt.figure(figsize=(8, 3))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original")

plt.subplot(1, 2, 2)
plt.imshow(inverted, cmap="gray")
plt.title("Inverted")

plt.show()
```
